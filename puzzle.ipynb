### Creating The Puzzle-15 and Board

The puzzle-15 is a 4x4 board containin  15 numbered tile and one *empty* tile.  
Each position on the board can be represented by an index:  

      0   1   2   3   
      4   5   6   7   
      8   *   *  *  
      *  *  *  *  
    
- "0" defines as the *empty* tile
- The board is made with a list in python

```python
goal_state = [
    0, 1, 2, 3,
    4, 5, 6, 7,
    8, -1, -1, -1,
    -1, -1, -1, -1
]


goal_state
```

```python
def print_board(board):
    for i in range(4):
        row = board[i*4:4*(i+1)]
        print(row)

print_board(goal_state)
```

To move any tile in the puzzle, we first need to know the position of the empty tile.

```python
def find_zero (board):
    return board.index(0)

find_zero (goal_state)
```

indetify the postion by the index:

```python
def index_position(x):
    row = x // 4
    column = x % 4
    return row, column

index_position (10)
```

**Moving the tiles using Swap function:**

```python
def swap (board, a, b):
    new_board = board.copy()
    new_board[a], new_board[b] = new_board[b], new_board[a]
    return new_board
move_1 = swap(goal_state, 10, 0)
print(move_1)
```

Since here, we can:
- Represent states
- Watch the board
- Find the empty tile
- Change index to position 
- Swap the tiles   

Althogh these components are the fundamental structure of the program, We have to determine the limitations, make valid and logical moves that are allowed.

We can only move the tiles that are next to the empty tile.
The possible moves are:
- up
- down
- left
- right  

However, not all of these moves are always legal.

```python
def get_neighbors(board):

    neighbors = []
    zero_index = find_zero(board)

    row, column = index_position(zero_index)

    if row > 0:
        new_board = swap(board, zero_index, zero_index - 4)
        neighbors.append(new_board)

    if row < 3:
        new_board = swap(board, zero_index, zero_index + 4)
        neighbors.append(new_board)

    if column > 0:
        new_board = swap(board, zero_index, zero_index - 1)
        neighbors.append(new_board)

    if column < 3:
        new_board = swap(board, zero_index, zero_index + 1)
        neighbors.append(new_board)

    return neighbors
```

Now, we evaluate this function.

```python
test_board = [
    1, 2, 3, 4, 
    6, 0, 5, -1,
    -1, -1, -1, -1, 
    8, -1, -1, 7
]

neighbors = get_neighbors(test_board)

neighbors
```

### Manhattan Distance (heuristic)
$
distance = |row1 - row2| + |col1 - col2|
$

- current position
- goal position
- calculating Manhattan distance 

```python
def normalize(board):

    return tuple(board)
```

```python
from collections import deque
```

```python
def explore_all_states(start_board):

    visited = set()
    queue = deque()

    start_tuple = normalize(start_board)

    visited.add(start_tuple)
    queue.append(start_board)

    expanded = 0

    while queue:

        current = queue.popleft()

        for neighbor in get_neighbors(current):

            norm = normalize(neighbor)

            if norm not in visited:
                visited.add(norm)
                queue.append(neighbor)

        expanded += 1
        if expanded % 1000 == 0:
            print("Expanded:", expanded)

    return visited
```

```python
print("Starting BFS ...")
all_states = explore_all_states(goal_state)
print("Finished ✅")
print("Total unique states:", len(all_states))
```

```python
import sqlite3
```

```python

conn = sqlite3.connect("puzzle.db")
c = conn.cursor()

c.execute("""
CREATE TABLE IF NOT EXISTS states (
    state TEXT PRIMARY KEY
)
""")

data = [(str(state),) for state in all_states]

c.executemany("INSERT OR IGNORE INTO states VALUES (?)", data)

conn.commit()
conn.close()

print("States saved to puzzle.db ✅")
```
